Importando bibliotecas

```python
import requests
import pandas as pd
from IPython.display import display
from dotenv import load_dotenv
import os
load_dotenv()
```

Código para chamar API

```python
api_key = os.environ["apiK"]

url = f"https://api.themoviedb.org/3/tv/top_rated?api_key={api_key}&amp;language=pt-BR"

response = requests.get(url)

data = response.json()
data
```

Consegui extrair os dados, agora vou consultar apenas o que eu tenho interesse

```python
filmes = []

for movie in data['results']:
    df = {'Título': movie['original_name'],
          'Data de Lançamento': movie['first_air_date'],
          'Visão geral': movie['overview'],
          'Votos': movie['vote_count'],
          'Média de votos': movie['vote_average']}
    
    filmes.append(df)

df = pd.DataFrame(filmes)
display(df)
          
```
