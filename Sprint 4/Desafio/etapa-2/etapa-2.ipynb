1 - Identifique a distribuição de óbitos por raça/cor (tp_raca_cor) em 2023 e calcule a porcentagem de cada grupo em relação ao total de óbitos.

Agrupei os óbitos por raça/cor.

Calculei a porcentagem de cada grupo.

```python
distribuicao_raca_cor = df['tp_raca_cor'].value_counts()

porcentagem_raca_cor = (distribuicao_raca_cor / distribuicao_raca_cor.sum()) * 100

print("Distribuição de óbitos por raça/cor em 2023:")
print(distribuicao_raca_cor)
print("\nPorcentagem de cada grupo:")
print(porcentagem_raca_cor.round(2))
```

2 - Classifique os óbitos em categorias de idade (ex: "0-18", "19-40", "41-60", "61+") e calcule a quantidade de óbitos em cada faixa etária.

Criei uma função condicional para classificar a idade (nu_idade) em faixas etárias.

Agrupei os óbitos por faixa etária e contei a quantidade em cada grupo.

```python
def classificar_idade(idade):
    if idade <= 18:
        return '0-18'
    elif 19 <= idade <= 40:
        return '19-40'
    elif 41 <= idade <= 60:
        return '41-60'
    else:
        return '61+'


df['Faixa_Etaria'] = df['nu_idade'].apply(classificar_idade)

distribuicao_faixa_etaria = df['Faixa_Etaria'].value_counts()

print("Distribuição de óbitos por faixa etária em 2023:")
print(distribuicao_faixa_etaria)
```

3 -  Identifique o mês com o maior número de óbitos em 2023 e calcule a média diária de óbitos nesse mês.


Extraí o mês da coluna dt_obito.

Agrupei os óbitos por mês e identifiquei o mês com maior ocorrência.

Calculei a média diária de óbitos no mês com maior número de óbitos.

```python
df['Mes'] = df['dt_obito'].str[3:5]

obitos_por_mes = df['Mes'].value_counts()
mes_maior_obitos = obitos_por_mes.idxmax()
total_obitos_mes = obitos_por_mes.max()

media_diaria = total_obitos_mes / 31

print("Mês com maior número de óbitos em 2023:", mes_maior_obitos)
print("Total de óbitos no mês:", total_obitos_mes)
print("Média diária de óbitos no mês:", round(media_diaria, 2))
```
